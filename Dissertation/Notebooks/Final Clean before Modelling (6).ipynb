```python
def add_wages(df, wages_df):    
    for team, wages, season in wages_df.values:
        #df = df.loc[df['season_x'] == season]
        df.loc[(df['home_team'] == team) & (df['season_x'] == season), 'home_team_wages'] = wages
        df.loc[(df['away_team'] == team) & (df['season_x'] == season), 'away_team_wages'] = wages
    return df
```

```python
df = add_wages(df, wage_df)
```
