```python
def rolling_averages(group, number, cols, new_cols):
    group = group.sort_values("date")
    rolling_stats = group[cols].rolling(number, closed='left').mean()
    group[new_cols] = rolling_stats
    return group
```

```python
number = 10
away_cols = ['away points','away constant points','away GF','away GA','away xG','away xGA', 'away adjusted GS', 'away adjusted GC']
away_cols_new = [f"away_team_{c}_rolling_{number}" for c in away_cols]
matches = matches.groupby("away_team").apply(lambda x: rolling_averages(x, number,away_cols, away_cols_new))
```

```python
home_cols = ['home points','home constant points','home GF','home GA','home xG','home xGA', 'home adjusted GS', 'home adjusted GC']
home_cols_new = [f"home_team_{c}_rolling_{number}" for c in home_cols]
matches = matches.groupby("home_team").apply(lambda x: rolling_averages(x, number,home_cols, home_cols_new))
```

```python
matches = matches.droplevel('away_team')
matches = matches.droplevel('home_team')
```

```python
matches.sort_index(inplace=True)
```
