```python
# Add Points and Goals Difference
df['gf'] = df['gf'].astype(float)
df['ga'] = df['ga'].astype(float)
df["xGD"] = df["xg"] - df["xga"]
df["GD"] = df["gf"] - df["ga"]
#df["GD"] = df["GD"].astype(int)

points = []
for x in range(len(df)):
    points.append(3 if df["result"][x] == "W" else 1 if df["result"][x] == "D" else 0 if df["result"][x] == "L" else float("nan"))

df["points"] = points
```

```python
def set_league_diff(df, point_weight, xGD_weight, GD_weight, GF_weight, GA_weight):   
    league_diff = []
    xGD_league_diff = []
    GD_league_diff = []
    GF_league_diff = []
    GA_league_diff = []
    xG_league_diff = []
    xGA_league_diff = []
    for x in range(len(df)):
        if df['comp'][x] == 'Championship':
            league_diff.append(point_weight)
            xGD_league_diff.append(xGD_weight)
            GD_league_diff.append(GD_weight)
            GF_league_diff.append(GF_weight)
            GA_league_diff.append(GA_weight)
            xG_league_diff.append(GF_weight)
            xGA_league_diff.append(GA_weight)
        else:
            league_diff.append(1)
            xGD_league_diff.append(0)
            GD_league_diff.append(0)
            GF_league_diff.append(0)
            GA_league_diff.append(0)
            xG_league_diff.append(0)
            xGA_league_diff.append(0)

    df["league_diff"] = league_diff
    df["xGD_league_diff"] = xGD_league_diff
    df["GD_league_diff"] = GD_league_diff
    df["GF_league_diff"] = GF_league_diff
    df["GA_league_diff"] = GA_league_diff
    df["xG_league_diff"] = xG_league_diff
    df["xGA_league_diff"] = xGA_league_diff
    return df

def cal_weighted_metrics(df):
    df["league_weighted_points"] = df["points"] * df["league_diff"]
    df["league_weighted_xGD"] = df["xGD"] + df["xGD_league_diff"]
    df["league_weighted_GD"] = df["GD"] + df["GD_league_diff"]
    df["league_weighted_GF"] = df["gf"] + df["GF_league_diff"]
    df["league_weighted_GA"] = df["ga"] + df["GA_league_diff"]
    df["league_weighted_xG"] = df["xg"] + df["xG_league_diff"]
    df["league_weighted_xGA"] = df["xga"] + df["xGA_league_diff"]
    return df
```

```python
point_weight = 0.30
xGD_weight = -1.25
GD_weight = -1.78
GF_weight = -0.76
GA_weight = +1.02
df = set_league_diff(df, point_weight, xGD_weight, GD_weight, GF_weight, GA_weight)
df = cal_weighted_metrics(df)
```

```python
def cal_season_points(df, season):
    # Filter the dataframe based on the season#
    team_points = defaultdict(int)
    team_xGD = defaultdict(int)
    df = df[df['season'] == season]

    # Iterate through the rows
    for _, row in df.iterrows():
        team = row['new_team']
        team_xGD[team] += (row['league_weighted_xGD'])
        result = row['result']
        league_diff = row['league_diff']

        if result == 'W':
            team_points[team] += (3 * league_diff)  # Home win, add 3 points to home team
        elif result == 'D':
            team_points[team] += (1 * league_diff)  # Draw, add 1 point to both teams
        else:
            team_points[team] += 0  # Away win, add 3 points to away team
    
    points_df = pd.DataFrame(list(team_points.items()), columns=['Team', 'Points'])
    xGD_df = pd.DataFrame(list(team_xGD.items()), columns=['Team', 'xGD'])
    new_df = pd.merge(points_df, xGD_df, on='Team')
    new_df = new_df.sort_values(by='Points', ascending=False)
    new_df = new_df.reset_index(drop=True)
    return new_df
```

```python
def cal_season_finish(df, season, league):
    # Filter the dataframe based on the season#
    team_points = defaultdict(int)
    team_xGD = defaultdict(int)
    team_GD = defaultdict(int)
    df = df[df['season'] == season]
    df = df[df['comp'] == league]

    # Iterate through the rows
    for _, row in df.iterrows():
        team = row['new_team']
        team_points[team] += (row['league_weighted_points'])
        team_xGD[team] += (row['league_weighted_xGD'])
        team_GD[team] += (row['league_weighted_GD'])
    
    points_df = pd.DataFrame(list(team_points.items()), columns=['Team', 'Points'])
    xGD_df = pd.DataFrame(list(team_xGD.items()), columns=['Team', 'xGD'])
    GD_df = pd.DataFrame(list(team_GD.items()), columns=['Team', 'GD'])
    new_df = pd.merge(points_df, xGD_df, on='Team')
    new_df = pd.merge(new_df, GD_df, on='Team')
    new_df = new_df.sort_values(by=['Points','GD'], ascending=False)
    if league == 'Premier League':
        new_df['Actual Position'] = range(1, len(new_df) + 1)
        new_df['Expected Position'] = new_df['xGD'].rank(ascending=False)
    elif league == 'Championship':
        new_df['Actual Position'] = range(21, len(new_df) + 21)
        new_df['Expected Position'] = (new_df['xGD'].rank(ascending=False)) + 20
    new_df['Points'] = points_df['Points']
    new_df['xGD'] = xGD_df['xGD']
    new_df['GD'] = GD_df['GD']
    new_df = new_df.sort_values(by='Actual Position', ascending=True)
    new_df = new_df.reset_index(drop=True)
    return new_df
```

```python
prem_2023 = cal_season_finish(df, 2023, 'Premier League')
prem_2022 = cal_season_finish(df, 2022, 'Premier League')
prem_2021 = cal_season_finish(df, 2021, 'Premier League')
prem_2020 = cal_season_finish(df, 2020, 'Premier League')
prem_2019 = cal_season_finish(df, 2019, 'Premier League')
prem_2018 = cal_season_finish(df, 2018, 'Premier League')
prem_2017 = cal_season_finish(df, 2017, 'Premier League')
champ_2023 = cal_season_finish(df, 2023, 'Championship')
champ_2022 = cal_season_finish(df, 2022, 'Championship')
champ_2021 = cal_season_finish(df, 2021, 'Championship')
champ_2020 = cal_season_finish(df, 2020, 'Championship')
champ_2019 = cal_season_finish(df, 2019, 'Championship')
champ_2018 = cal_season_finish(df, 2018, 'Championship')
champ_2017 = cal_season_finish(df, 2017, 'Championship')
points_2017 = pd.concat([prem_2017, champ_2017])
points_2018 = pd.concat([prem_2018, champ_2018])
points_2019 = pd.concat([prem_2019, champ_2019])
points_2020 = pd.concat([prem_2020, champ_2020])
points_2021 = pd.concat([prem_2021, champ_2021])
points_2022 = pd.concat([prem_2022, champ_2022])
points_2023 = pd.concat([prem_2023, champ_2023])
```

```python
def add_previous_points(df, season, point_df):
    for team, points, xGD, GD, position, xposition in point_df.values:
        df = df.loc[df['season'] == season]
        df.loc[df['new_team'] == team, 'new_team_previous_points'] = points
        df.loc[df['new_team'] == team, 'new_team_previous_xGD'] = xGD
        df.loc[df['new_team'] == team, 'new_team_previous_GD'] = GD
        df.loc[df['new_team'] == team, 'new_team_previous_position'] = position
        df.loc[df['new_team'] == team, 'new_team_previous_expected_position'] = xposition
    return df
```

```python
df2017 = df[df['season'] == 2017]
df2018 = add_previous_points(df, 2018, points_2017)
df2019 = add_previous_points(df, 2019, points_2018)
df2020 = add_previous_points(df, 2020, points_2019)
df2021 = add_previous_points(df, 2021, points_2020)
df2022 = add_previous_points(df, 2022, points_2021)
df2023 = add_previous_points(df, 2023, points_2022)
df2024 = add_previous_points(df, 2024, points_2023)
```
