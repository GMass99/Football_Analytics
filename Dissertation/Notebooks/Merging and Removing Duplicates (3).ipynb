```python
merged.sort_values(["date","home_team"], inplace=True)
merged.reset_index(drop=True, inplace=True)
for x in range(len(merged)-1):
    if merged["date"][x] == merged["date"][x+1] and merged["new_team_x"][x] == merged["new_team_y"][x+1]: 
        merged.drop(index=x, inplace=True)
    else:
        continue
merged.reset_index(drop=True, inplace=True)
```
