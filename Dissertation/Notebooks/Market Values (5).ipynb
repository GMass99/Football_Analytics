```python
def add_market_value(df, market_value_df):
    for team, season, mvalue in market_value_df.values:
        #df = df.loc[df['season_x'] == season]
        df.loc[(df['home_team'] == team) & (df['season_x'] == season), 'home_team_market_value'] = mvalue
        df.loc[(df['away_team'] == team) & (df['season_x'] == season), 'away_team_market_value'] = mvalue
    return df
```

```python
df = add_market_value(matches, market_value_df)
```
